```python
import numpy as np
import pandas as pd
import matplotlib.pyplot as plt
```

```python
data ={
    'Name':['A','B','C','D','E'],
    'Math':[35,56,67,89,78],
    'Science':[67,98,54,27,32],
    'English':[12,45,68,54,31]
}
df=pd.DataFrame(data)
print(df)
```

```python
marks=df[["Math","Science","English"]].values
average_marks=np.mean(marks,axis=1)
df["Average"]= average_marks
print(df)
```

```python
def grade(avg):
  if avg >= 90:
    return 'A'
  elif avg >= 75:
    return 'B'
  elif avg >= 60:
    return 'C'
  else:
    return 'D'
df['Grade']=df ["Average"].apply(grade)
```

```python
plt.figure()
plt.bar(df["Name"],df["Average"])
plt.xlabel('Students')
plt.ylabel('Average marks')
plt.title('Average marks of students ')
plt.show()
```

```python
df[["Math","Science","English"]].mean().plot(kind="bar")
plt.title("Average marks for subject")
plt.ylabel("Marks")
plt.show()
```
